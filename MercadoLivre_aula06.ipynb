# Web scraping com Python
# Buscando produtos no Mercado Livre

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
```

```python
url_base = 'https://lista.mercadolivre.com.br/'
```

```python
produto_nome = input('Qual produto você procura? ')
```

```python
response = requests.get(url_base + produto_nome)
print(response)
```

```python
response.text
```

```python
site = BeautifulSoup(response.text, 'html.parser') #faz a requisição passando  o  formato html
```

```python
site
```

```python
#procurar pela tag div e os atributos attrs

produto = site.find('div', attrs= {'class' : 'ui-search-result__wrapper'})
```

```python
print(produto)
```

```python
links = produto.find('a', attrs={'class', 'ui-search-link'})
```

```python
print('Link do produto: ', links['href'])
```

```python
#localizando o preço

real= produto.find('span', attrs={'class': 'andes-money-amount__fraction'})
preco = real.text.split()[0]
print(preco)
```

```python
#localizando o preço em outra ocorrencia.

real= produto.findAll('span', attrs={'class': 'andes-money-amount__fraction'})
preco = real[1]
print(preco.text)
```

```python
print('Preço do produto: R$ ', preco)
```

# Buscando todos os produtos

```python
url_base = 'https://lista.mercadolivre.com.br/'
```

```python
#criando um script

from numpy.lib.function_base import append
produto_nome=input('qual produto você procura? ')
response = requests.get(url_base + produto_nome)
site = BeautifulSoup(response.text, 'html.parser')
produtos= site.findAll('div', attrs= {'class':'ui-search-result__wrapper'})


a = []
for produto in produtos:

  links = produto.find('a', attrs={'class', 'ui-search-link'})
  real= produto.find('span', attrs={'class': 'andes-money-amount__fraction'})
  preco = real.text.split()[0]
  print('Link do produto: ', links['href'])
  print('Preço do produto: R$ ', preco)
  valor= preco
  a+=[valor]

  print('\n')
```

```python
print(a)
```

```python
dados = pd.DataFrame(a)
dados.columns = ['Preço']
```

```python
dados
```

```python
dados['Preço'] = pd.to_numeric(dados['Preço'])
dados.head()
```
